## ⚡ 3. Shortcut iOS (opțional, pentru workflow rapid)

Ca să nu mai deschizi Juno manual de fiecare dată:

**Crează un iOS Shortcut:**
1. Aplicația **Shortcuts** → **+** → New Shortcut
2. Adaugă acțiunea: **Take Screenshot**
3. Adaugă: **Save to Files** → folder Juno → salvează ca `latest.png`
4. Adaugă: **Open App** → Juno
5. Adaugă shortcut-ul la **Back Tap** (Settings → Accessibility → Touch → Back Tap → Triple Tap)

Acum: **3 tap-uri pe spatele iPhone-ului** → screenshot salvat → deschide Juno → rulezi celula de analiză.
